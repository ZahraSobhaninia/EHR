```python
def compare_patient(p1, p2):
    result = {}

    result["outcome"]  = p1.outcome == p2.outcome
    result["concepts"] = p1.concepts == p2.concepts
    result["segments"] = p1.segments == p2.segments
    result["ages"]     = p1.ages == p2.ages
    result["abspos"]   = p1.abspos == p2.abspos

    result["all_equal"] = all(result.values())

    return result
```

```python
def compare_datasets(map1, map2, name1="A", name2="B"):
    common_pids = set(map1) & set(map2)

    print(f"\nComparing {name1} vs {name2}")
    print("Common patients:", len(common_pids))

    identical = 0
    different = 0

    diffs = []

    for pid in common_pids:
        res = compare_patient(map1[pid], map2[pid])

        if res["all_equal"]:
            identical += 1
        else:
            different += 1
            diffs.append((pid, res))

    print("Identical:", identical)
    print("Different:", different)

    return diffs
```

```python
def compare_sequences(p1, p2):
    return p1.concepts == p2.concepts
```

```python
def compare_sequence_only(map1, map2, name1="A", name2="B"):
    common_pids = set(map1) & set(map2)

    same = 0
    diff = 0

    for pid in common_pids:
        if map1[pid].concepts == map2[pid].concepts:
            same += 1
        else:
            diff += 1

    print(f"\n{name1} vs {name2}")
    print("Same sequences:", same)
    print("Different sequences:", diff)
```

```python
def debug_sequence(pid, map1, map2):
    p1 = map1[pid]
    p2 = map2[pid]

    c1 = p1.concepts
    c2 = p2.concepts

    print("Length:", len(c1), len(c2))

    for i, (a, b) in enumerate(zip(c1, c2)):
        if a != b:
            print(f"First difference at index {i}: {a} vs {b}")
            break
```
